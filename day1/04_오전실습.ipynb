## 미션 4. 함수로 반복 로직 묶기 🔧

> **상황:** API 호출은 매번 같은 구조를 반복합니다.
> 함수로 묶으면 코드 중복을 없애고, 수정 시 한 곳만 바꾸면 됩니다.

```python
def 함수명(인자: 타입 = 기본값) -> 반환타입:
    """독스트링"""
    return 결과
```

````python
# ─── 미션 4-1. 기본 함수 정의 ────────────────────────────────────
def greet_employee(name: str) -> str:
    """직원 이름을 받아 환영 메시지를 반환한다."""
    return f"안녕하세요, {name}님! 노바코프 HR 챗봇입니다."

# 함수 호출 테스트
print(greet_employee("김인턴"))
print(greet_employee("이팀장"))
````

````python
# ─── 미션 4-2. 기본값이 있는 함수 ───────────────────────────────
def call_hr_llm(
    prompt: str,
    temperature: ___ = ___,   # float, 기본값 0.3
    max_tokens:  ___ = ___    # int, 기본값 512
) -> str:
    """HR 챗봇 LLM 호출 함수 템플릿"""
    return f"[temp={temperature}, max={max_tokens}] {prompt}"

print("=== 다양한 호출 방식 ===")

# 기본값 사용
print(call_hr_llm("연차 사용법 알려줘"))

# temperature만 변경 (키워드 인자)
print(call_hr_llm("연차 사용법 알려줘", temperature=0.0))

# max_tokens만 변경
print(call_hr_llm("연차 사용법 알려줘", max_tokens=100))

# 둘 다 변경
print(call_hr_llm("연차 사용법 알려줘", 0.0, 256))
````

````python
# ─── 미션 4-3. 반환값 없는 함수 vs 있는 함수 ─────────────────────

# 반환값 없음 — 출력만 함
def print_hr_response(text: str) -> None:
    print(f"HR봇: {text}")

# 반환값 있음 — 결과를 변수에 저장 가능
def get_hr_response(text: str) -> str:
    return f"HR봇: {text}"

# None 반환 — 저장 불가
print_hr_response("연차는 1년에 15일입니다.")

# str 반환 — 저장 후 추가 처리 가능
response = get_hr_response("연차는 1년에 15일입니다.")
print(response)
print(f"응답 길이: {len(response)}자")
````

## 미션 6. f-string으로 프롬프트 만들기 ✍️

> **상황:** HR 챗봇의 프롬프트를 동적으로 생성합니다.

| 패턴 | 용도 |
|------|------|
| `f"{변수}"` | 변수 삽입 |

| `f"{숫자:,}"` | 천 단위 쉼표 |
| `"""멀티라인""".strip()` | 긴 프롬프트 작성 |

| 함수 | 방향 | 암기 팁 |
|------|------|---------|
| `json.loads(str)` | 문자열 → 딕셔너리 | s = string |
| `json.dumps(dict)` | 딕셔너리 → 문자열 | s = string |
| `json.load(file)` | 파일 → 딕셔너리 | |
| `json.dump(dict, file)` | 딕셔너리 → 파일 | |

````python
# ─── 미션 7-1. json.loads() — 문자열 → 딕셔너리 ──────────────────
import json

# LLM이 돌려준 JSON 응답 (문자열)
llm_response = '{"score": 4, "reason": "규정 조항 포함", "missing": ["예시"]}'

print(f"원본 타입: {type(llm_response)}")

# 딕셔너리로 변환
parsed = json.___(llm_response)

print(f"변환 후 타입: {type(parsed)}")
print(f"score:   {parsed[___]}")
print(f"reason:  {parsed['reason']}")
print(f"missing: {parsed['missing']}")
print(f"첫 번째: {parsed['missing'][0]}")
````

````python
# ─── 미션 7-2. json.dumps() — 딕셔너리 → 문자열 ──────────────────
hr_data = {
    "instruction": "연차 신청은 어떻게 하나요?",
    "input"      : "",
    "output"     : "HR 포털 → 휴가 신청 → 연차 선택 후 제출"
}

# 한글 깨짐 버전 (기본값 ensure_ascii=True)
broken = json.dumps(hr_data)
print("⚠️ 한글 깨짐:")
print(broken[:60] + "...")

print()

# 한글 정상 버전
good = json.___(hr_data, ensure_ascii=___, indent=2)
print("✅ 한글 정상 (ensure_ascii=False, indent=2):")
print(good)
````
