````python
# ✅ MISSION 2 정답 ─ 자유 주제 호출 (예시: 이탈리아 셰프)
response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {"role": "system", "content": "당신은 이탈리아 요리 전문 셰프입니다."},   # 전문가 역할 설정
        {"role": "user",   "content": "파스타를 맛있게 삶는 방법을 알려주세요."},  # 질문
    ],
    temperature=0.7,   # 0.5 이상
    max_tokens=200,    # 300 이하
)

print(response.choices[0].message.content)
````

````python
# ✅ MISSION 3 정답 ─ temperature 비교
response_low = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[{"role": "user", "content": "오늘 점심 메뉴를 추천해줘"}],
    temperature=0.1,   # 낮음 → 일정한 답변
    max_tokens=100,
)

response_high = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[{"role": "user", "content": "오늘 점심 메뉴를 추천해줘"}],
    temperature=1.5,   # 높음 → 다양한 답변
    max_tokens=100,
)

print("🧊 temperature=0.1 (낮음):")
print(response_low.choices[0].message.content)   # choices[0].message.content 로 꺼냄

print("\n🔥 temperature=1.5 (높음):")
print(response_high.choices[0].message.content)
````

````python
# ✅ MISSION 4 정답 ─ 응답 객체 값 추출
response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[{"role": "user", "content": "Python의 장점을 두 가지만 알려줘"}],
    temperature=0.7,
    max_tokens=150,
)

print("📝 답변:")
print(response.choices[0].message.content)        # ① 답변 텍스트

print("\n🏁 종료 이유:", response.choices[0].finish_reason)   # ② 종료 이유

print("\n📊 입력 토큰:", response.usage.prompt_tokens)         # ③ 입력 토큰
print("📊 출력 토큰:", response.usage.completion_tokens)       # ④ 출력 토큰
````

````python
# ✅ MISSION 5 정답 ─ 노바코프 역할극 챗봇
system_msg = "당신은 노바코프 인사팀 담당자입니다. 직원들의 질문에 친절하고 간결하게 답변하세요."

r1 = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {"role": "system", "content": system_msg},
        {"role": "user",   "content": "연차는 몇 일이나 되나요?"},   # 질문 1
    ],
    temperature=0.5,
    max_tokens=150,
)

r2 = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {"role": "system", "content": system_msg},
        {"role": "user",   "content": "재택근무는 주에 며칠 가능한가요?"},   # 질문 2
    ],
    temperature=0.5,
    max_tokens=150,
)

r3 = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {"role": "system", "content": system_msg},
        {"role": "user",   "content": "결혼 축하금은 얼마인가요?"},   # 질문 3
    ],
    temperature=0.5,
    max_tokens=150,
)

print("Q1 답변:", r1.choices[0].message.content)
print("\nQ2 답변:", r2.choices[0].message.content)
print("\nQ3 답변:", r3.choices[0].message.content)
````

````python
# ✅ MISSION 6 정답 ─ 비용 계산기
total_input  = r1.usage.prompt_tokens + r2.usage.prompt_tokens + r3.usage.prompt_tokens        # 입력 합계
total_output = r1.usage.completion_tokens + r2.usage.completion_tokens + r3.usage.completion_tokens  # 출력 합계

# 비용 계산: (입력×0.15 + 출력×0.60) ÷ 1,000,000
total_cost = (total_input * 0.15 + total_output * 0.60) / 1_000_000

print(f"총 입력 토큰: {total_input}")
print(f"총 출력 토큰: {total_output}")
print(f"총 비용: ${total_cost:.6f}  (약 {total_cost * 1400:.2f}원)")  # 1달러 = 1,400원
````
