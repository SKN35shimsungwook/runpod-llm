## 1. 파이썬 기본 자료형 — LLM 코드에서 만나는 4가지 <a id="1"></a>

파이썬에는 여러 자료형이 있지만, LLM 코드에서 핵심적으로 사용하는 것은 4가지입니다.

| 자료형 | 파이썬 표기 | 예시 | LLM 코드에서의 역할 |
|--------|------------|------|--------------------|
| **문자열** (str) | `"텍스트"` | `"gpt-4o-mini"` | 모델명, 프롬프트, 응답 텍스트 |
| **정수** (int) | `숫자` | `512` | max_tokens, 반복 횟수 |
| **실수** (float) | `소수점` | `0.7` | temperature, top_p |
| **불리언** (bool) | `True/False` | `True` | GPU 연결 여부, 조건 판단 |

````python
# ── 1-1. LLM 코드에서 만나는 자료형들 ──────────────────────────────
model_name    = "gpt-4o-mini"   # str   — 모델명
temperature   = 0.7              # float — 창의성 조절
max_tokens    = 512              # int   — 최대 출력 길이
gpu_available = True             # bool  — GPU 연결 여부

print("=== 자료형 확인 ===")
print(f"model_name    = {model_name!r:20} | type: {type(model_name).__name__}")
print(f"temperature   = {temperature!r:20} | type: {type(temperature).__name__}")
print(f"max_tokens    = {max_tokens!r:20} | type: {type(max_tokens).__name__}")
print(f"gpu_available = {gpu_available!r:20} | type: {type(gpu_available).__name__}")
````

````python
# ── 1-2. type() 함수로 자료형 확인 ───────────────────────────────────
print(type(model_name))     # <class 'str'>
print(type(temperature))    # <class 'float'>
print(type(max_tokens))     # <class 'int'>
print(type(gpu_available))  # <class 'bool'>
````

### 📋 리스트 — 여러 값을 순서대로 담기

리스트는 대괄호 `[]` 안에 여러 값을 담는 자료형입니다.  
API의 `messages`가 바로 리스트 구조입니다.

````python
# ── 1-3. 리스트 기본 조작 ──────────────────────────────────────────
topics = ["임대차 분쟁", "근로계약 해지", "소비자 분쟁"]

print("=== 리스트 기본 조작 ===")
print(f"topics[0]   = {topics[0]}")   # 첫 번째 — 인덱스는 0부터!
print(f"topics[1]   = {topics[1]}")   # 두 번째
print(f"topics[-1]  = {topics[-1]}")  # 마지막 (음수 인덱스)
print(f"len(topics) = {len(topics)}") # 리스트 길이

# 값 추가
topics.append("상속 문제")
print(f"\n추가 후: {topics}")
print(f"len(topics) = {len(topics)}")
````

````python
# ── 1-4. 리스트 반복 — Day 2 합성 데이터 생성에서 사용 ─────────────
print("=== for 루프로 리스트 반복 ===")

for topic in topics:
    print(f"  주제: {topic}")

# 인덱스와 함께 반복
print("\n=== 인덱스 포함 반복 ===")
for i, topic in enumerate(topics):
    print(f"  [{i}] {topic}")
````

## 3. 함수 — 반복 호출 로직을 하나로 묶기 <a id="3"></a>

API 호출은 매번 같은 구조를 반복합니다.  
이 로직을 **함수로 묶으면** 코드 중복을 없애고, 수정이 필요할 때 한 곳만 바꾸면 됩니다.

````python
# ── 3-1. 함수 기본 구조 ──────────────────────────────────────────────
def greet(name: str) -> str:
    """이름을 받아 인사말을 반환한다."""
    return f"안녕하세요, {name}님!"

print(greet("진환"))   # "안녕하세요, 진환님!"
print(greet("여러분")) # "안녕하세요, 여러분님!"
````

함수 정의 부분의 끝(:) 바로 앞에 화살표(->)와 자료형을 적어줍니다.

def 함수명(...) -> str: : 이 함수는 문자열을 반환한다.

def 함수명(...) -> int: : 이 함수는 정수를 반환한다.

def 함수명(...) -> None: : 이 함수는 반환값(return)이 없다.

````python
# ── 3-2. 타입 힌트 — 있을 때 vs 없을 때 비교 ────────────────────────

# 타입 힌트 없는 버전
def call_llm_v1(prompt, temperature):
    result = "응답 예시"
    return result

# 타입 힌트 있는 버전 — 훨씬 명확함
def call_llm_v2(prompt: str, temperature: float = 0.7) -> str:
    """LLM을 호출하고 응답 텍스트를 반환한다.

    Args:
        prompt: 사용자 입력 프롬프트
        temperature: 응답 다양성 조절 (0.0~2.0)

    Returns:
        LLM 응답 텍스트
    """
    result = f"[temperature={temperature}] {prompt}에 대한 응답"
    return result

print("=== call_llm_v2 테스트 ===")
print(call_llm_v2("계약서 검토"))
print(call_llm_v2("계약서 검토", 0.0))
print(call_llm_v2("판결문 분석", temperature=0.2))
````

````python
# ── 3-3. 기본값 매개변수 — 다양한 호출 방식 ─────────────────────────
def call_llm(
    prompt: str,
    temperature: float = 0.7,
    max_tokens: int = 512
) -> str:
    """LLM 호출 함수 템플릿 — 오후에 실제 API로 채울 껍데기"""
    return f"[temp={temperature}, max={max_tokens}] → {prompt[:20]}..."

print("=== 기본값 매개변수 호출 방식 ===")
print(call_llm("계약서 검토"))                           # 모든 기본값 사용
print(call_llm("계약서 검토", 0.0))                      # temperature만 변경
print(call_llm("계약서 검토", temperature=0.0))           # 키워드 인자로 명시
print(call_llm("계약서 검토", max_tokens=100))            # max_tokens만 변경
print(call_llm("계약서 검토", 0.0, 256))                  # 둘 다 변경
````

````python
# ── 3-4. 반환값 없는 함수 vs 있는 함수 ──────────────────────────────

# 반환값 없음 — 출력만 함
def print_result(text: str) -> None:
    print(f"결과: {text}")

# 반환값 있음 — 결과를 변수에 저장 가능 (LLM 코드에서 주로 사용)
def get_result(text: str) -> str:
    return f"결과: {text}"

print("=== 반환값 비교 ===")
print_result("테스트")        # 화면에 바로 출력, 저장 불가

result = get_result("테스트") # 변수에 저장
print(result)                  # 원할 때 출력
print(f"결과 길이: {len(result)}")  # 추가 처리 가능

print()
print("💡 LLM 코드에서는 반환값이 있는 함수를 주로 사용합니다.")
print("   API 응답을 받아서 다음 처리(저장, 평가, 표시)에 넘겨야 하기 때문입니다.")
````

## 4. 딕셔너리 — API 요청/응답 구조와 1:1 대응 <a id="4"></a>

딕셔너리는 **"키: 값"** 쌍으로 데이터를 저장하는 구조입니다.  
> 💡 **핵심:** OpenAI API에 보내는 JSON이 곧 파이썬 딕셔너리입니다.

````python
# ── 4-1. 딕셔너리 기본 ───────────────────────────────────────────────
person = {
    "name" : "김진환",
    "role" : "데이터 사이언티스트",
    "years": 8
}

print("=== 딕셔너리 기본 접근 ===")
print(f"name  = {person['name']}")   # 문자열
print(f"role  = {person['role']}")
print(f"years = {person['years']}")  # 정수
print(f"type  = {type(person)}")
````

````python
# ── 4-2. API 요청 구조 = 딕셔너리 ───────────────────────────────────
#
# 실제 API에 보내는 요청 본문 — 이 구조를 오후에 그대로 씁니다
#
request_body = {
    "model": "gpt-4o-mini",
    "messages": [                           # ← 리스트 안에 딕셔너리들
        {"role": "system", "content": "당신은 법률 전문가입니다."},
        {"role": "user",   "content": "계약서 검토를 도와주세요."}
    ],
    "temperature": 0.3,
    "max_tokens" : 512
}

print("=== API 요청 구조 확인 ===")
print(f"모델명    : {request_body['model']}")
print(f"온도값    : {request_body['temperature']}")
print(f"최대토큰  : {request_body['max_tokens']}")
print(f"메시지 수 : {len(request_body['messages'])}개")
````

````python
# 첫 번째 메시지(system)의 content
print(request_body["messages"][0]["content"])

# 두 번째 메시지(user)의 content
print(request_body["messages"][1]["content"])
````

````python
print("--- 모든 메시지 내용 출력 ---")
for msg in request_body["messages"]:
    print(msg["content"])
````

````python
# ── 4-3. 딕셔너리 핵심 조작 4가지 ───────────────────────────────────
print("=== 1. 값 읽기 ([] 사용) ===")
print(request_body["model"])          # "gpt-4o-mini"
print(request_body["temperature"])    # 0.3

print()
print("=== 2. 값 수정 ===")
request_body["temperature"] = 0.7    # 값 변경
print(f"변경 후 temperature = {request_body['temperature']}")  # 0.7

print()
print("=== 3. 새 키 추가 ===")
request_body["top_p"] = 0.9          # 없는 키 → 새로 추가
print(f"추가된 top_p = {request_body['top_p']}")

print()
print("=== 4. 키 존재 확인 ===")
print(f"'temperature' in request_body = {'temperature' in request_body}")
print(f"'top_k' in request_body       = {'top_k' in request_body}")
````

````python
print("--- 대화 내역 확인 ---")
for msg in request_body["messages"]:
    role = msg["role"]
    content = msg["content"]
    print(f"[{role}]: {content}")
````

값 = 딕셔너리.get(찾으려는_키, 키가_없을_때_대신_내보낼_값)

````python
# ── 4-4. [] vs .get() 차이 ───────────────────────────────────────────
print("=== [] vs .get() 비교 ===")

# ✅ .get() — 키 없어도 안전
print(f".get('stream', False)      = {request_body.get('stream', False)}")  # 기본값 반환
print(f".get('model', '없음')     = {request_body.get('model', '없음')}")   # 실제 값
print(f".get('top_k', '설정 안됨') = {request_body.get('top_k', '설정 안됨')}")

print()
print("=== ❌ [] 로 없는 키 접근 시 KeyError 발생 ===")
try:
    _ = request_body["stream"]      # 없는 키
except KeyError as e:
    print(f"KeyError 발생: {e} ← 이래서 .get()을 사용!")
````

### 🔗 중첩 딕셔너리 접근 — 단계별 분해

API 요청/응답에서 가장 많이 만나는 구조가 **"리스트 안에 딕셔너리"** 입니다.

```
request_body["messages"][0]["role"]
      │           │      │     │
      │           │      │     └─ 딕셔너리의 "role" 키
      │           │      └─────── 리스트의 첫 번째 요소 (인덱스 0)
      │           └────────────── "messages" 키의 값 (리스트)
      └────────────────────────── 최상위 딕셔너리
```

````python
# ── 4-5. 중첩 딕셔너리 — 단계별로 쪼개서 이해 ──────────────────────
print("=== 단계별 분해 ===")

# 1단계: messages 키 → 리스트
messages = request_body["messages"]
print(f"1단계 | type: {type(messages).__name__} | 길이: {len(messages)}")

# 2단계: [0] → 첫 번째 딕셔너리
first_message = messages[0]
print(f"2단계 | type: {type(first_message).__name__} | 내용: {first_message}")

# 3단계: ["role"] → 문자열 값
role = first_message["role"]
print(f"3단계 | type: {type(role).__name__} | 값: {role!r}")

print()
print("=== 한 줄로 쓰기 (위와 동일한 결과) ===")
print(f"system role    = {request_body['messages'][0]['role']!r}")
print(f"user content   = {request_body['messages'][1]['content']!r}")
````

````python
# ── 4-6. 흔한 실수 — [0] 인덱스 누락 ────────────────────────────────
print("=== ⚠️ 흔한 실수: [0] 누락 ===")

try:
    # ❌ 잘못된 접근: 리스트에 바로 키로 접근
    wrong = request_body["messages"]["role"]
except TypeError as e:
    print(f"TypeError 발생: {e}")
    print("→ 리스트 안에 있으므로 반드시 [0]으로 먼저 접근해야 합니다!")

print()
print("=== ✅ 올바른 접근 ===")
correct = request_body["messages"][0]["role"]   # [0] 먼저!
print(f"올바른 접근: {correct!r}")
````

````python
# ── 4-7. 딕셔너리 리스트 반복 — Day 3 파이프라인에서 사용 ───────────
documents = [
    {"id": 1, "text": "원고는 피고에게 손해배상을 청구하였다.",    "category": "민사"},
    {"id": 2, "text": "피의자는 묵비권을 행사할 수 있다.",         "category": "형사"},
    {"id": 3, "text": "임차인은 보증금 반환을 요구할 권리가 있다.", "category": "민사"},
]

print("=== 딕셔너리 리스트 반복 ===")
for doc in documents:
    print(f"[{doc['category']}] ID:{doc['id']} | {doc['text']}")
````

## 5. 딕셔너리와 JSON의 관계 <a id="5"></a>

딕셔너리와 JSON은 생김새가 거의 같습니다.  
API는 **JSON 형식으로 통신**하고, 파이썬은 **딕셔너리로 처리**합니다.

| 파이썬 | JSON | 주의사항 |
|--------|------|----------|
| `{}` | `{}` | 키는 반드시 문자열 |
| `[]` | `[]` | 동일 |
| `True / False` | `true / false` | 대소문자 다름 |
| `None` | `null` | 이름 다름 |

````python
# ── 5-1. json 모듈 — 딕셔너리 ↔ 문자열 변환 ─────────────────────────
import json

# API 응답으로 받은 JSON 문자열 예시 (Judge 평가 결과)
raw_response = '{"accuracy": 4, "clarity": 5, "total": 9, "reason": "근거가 명확함"}'

print("=== JSON 문자열 → 딕셔너리 (json.loads) ===")
parsed = json.loads(raw_response)       # 문자열 → 딕셔너리
print(f"type    : {type(parsed)}")
print(f"accuracy: {parsed['accuracy']}")
print(f"reason  : {parsed['reason']}")
````

````python
# ── 5-2. 딕셔너리 → JSON 문자열 (json.dumps) ─────────────────────────
data = {
    "instruction": "임차인이 보증금을 돌려받지 못할 때 어떻게 해야 하나요?",
    "input"      : "",
    "output"     : "내용증명 발송 후 임차권등기명령을 신청하세요."
}

print("=== 딕셔너리 → JSON 문자열 (json.dumps) ===")
json_str = json.dumps(data, ensure_ascii=False, indent=2)
print(json_str)

print()
print("⚠️ ensure_ascii=False 없으면 한글이 깨집니다!")
broken = json.dumps(data, indent=2)  # ensure_ascii=True (기본값)
print(broken[:80] + "...")           # 한글이 \uXXXX로 변환됨
````

````python
# ── 5-3. JSON 4가지 함수 정리 ────────────────────────────────────────
print("=== JSON 4가지 함수 ===")
print()
print("json.loads(str)       → 문자열 → 딕셔너리  (s = string)")
print("json.dumps(dict)      → 딕셔너리 → 문자열  (s = string)")
print("json.load(file)       → 파일    → 딕셔너리")
print("json.dump(dict, file) → 딕셔너리 → 파일")
print()
print("💡 암기 팁: s가 붙으면 문자열(string) 대상, 없으면 파일 대상")

# 파일 저장 예시
sample_data = [
    {"id": 1, "text": "예시 텍스트", "label": "민사"},
    {"id": 2, "text": "두 번째 텍스트", "label": "형사"},
]

output_path = "/content/sample_output.json"
with open(output_path, "w", encoding="utf-8") as f:
    json.dump(sample_data, f, ensure_ascii=False, indent=2)

print(f"\n✅ 파일 저장 완료: {output_path}")

# 다시 읽기
with open(output_path, "r", encoding="utf-8") as f:
    loaded = json.load(f)

print(f"✅ 파일 읽기 완료: {len(loaded)}건")
````
