## 2. GPU 연결 확인 <a id="2"></a>

### ⚠️ 먼저 T4 GPU 설정을 했나요?

```
상단 메뉴 → 런타임 → 런타임 유형 변경 → T4 GPU → 저장
```

설정했으면 아래 셀을 실행하세요.

````python
# ✋ 체크포인트 2 — 전원 True 출력을 확인하세요
import torch

# ① GPU 연결 여부
is_available = torch.cuda.is_available()
print(f"GPU 연결 여부: {is_available}")

if is_available:
    # ② GPU 이름
    gpu_name = torch.cuda.get_device_name(0)
    print(f"GPU 이름:     {gpu_name}")

    # ③ VRAM 크기
    total_mem = torch.cuda.get_device_properties(0).total_memory / (1024**3)
    print(f"VRAM 크기:    {total_mem:.2f} GB")

    print()
    print("✅ T4 GPU 정상 연결!")
else:
    print()
    print("❌ GPU가 연결되지 않았습니다.")
    print("   해결 방법: 런타임 → 런타임 유형 변경 → T4 GPU → 저장")
````

## 3. T4 메모리 제약과 실습 가능 범위 <a id="3"></a>

T4 VRAM은 약 15GB입니다. 이번 과정에서 쓸 2–3B 모델은 안정적으로 실습할 수 있어요.

| 모델 크기 | 양자화 | VRAM 사용량 | T4 가능? |
|-----------|--------|------------|----------|
| 7B | FP16 | 약 14GB | ⚠️ 불안정 |
| 7B | 4-bit | 약 5–6GB | ✅ 가능 |
| **2–3B** | **FP16** | **약 5–6GB** | **✅ 가능** |
| **2–3B** | **4-bit** | **약 2–3GB** | **✅ 매우 안정** |

````python
# VRAM 사용량 실시간 확인
if torch.cuda.is_available():
    allocated = torch.cuda.memory_allocated() / 1e9
    reserved  = torch.cuda.memory_reserved()  / 1e9
    total     = torch.cuda.get_device_properties(0).total_memory / 1e9
    free      = total - reserved

    print("=== VRAM 현황 ===")
    print(f"  전체:  {total:.1f} GB")
    print(f"  사용:  {allocated:.1f} GB")
    print(f"  예약:  {reserved:.1f} GB")
    print(f"  여유:  {free:.1f} GB")
else:
    print("GPU가 연결되지 않아 확인할 수 없습니다.")
````

````python
# OOM(메모리 부족) 발생 시 대처법
print("=== CUDA out of memory 에러 발생 시 ===")
print()
print("1. batch_size 줄이기: 4 → 2 → 1")
print("2. 아래 코드로 메모리 강제 해제:")
print("     import gc")
print("     del model")
print("     gc.collect()")
print("     torch.cuda.empty_cache()")
print("3. 그래도 안 되면: 런타임 → 런타임 다시 시작")
print()
print("💡 Day 4에서 양자화로 메모리를 절약하는 방법을 배웁니다!")
````

````python
# 에러 1: ModuleNotFoundError — 패키지 미설치
print("=== 에러 1: ModuleNotFoundError ===")
try:
    import gradio   # 아직 설치 안 함
except ModuleNotFoundError as e:
    print(f"❌ {e}")
    print("   해결: !pip install -q gradio")
    print("   원인: 패키지 미설치 또는 세션 재시작 후 재설치 안 함")
````

````python
# 에러 2: NameError — 변수를 정의한 셀을 실행하지 않음
print("=== 에러 2: NameError ===")
try:
    print(undefined_variable)
except NameError as e:
    print(f"❌ {e}")
    print("   해결: 위쪽 셀부터 순서대로 실행 (Ctrl + F9)")
    print("   원인: 변수를 정의한 셀을 아직 실행하지 않음")
````

````python
# 에러 3: JSONDecodeError — LLM 응답이 JSON 형식이 아닐 때
print("=== 에러 3: JSONDecodeError ===")
try:
    json.loads("이것은 JSON이 아닙니다")
except json.JSONDecodeError as e:
    print(f"❌ {e}")
    print("   해결: 프롬프트에 'JSON만 출력' 강조")
    print("         .replace('```json', '').replace('```', '').strip() 전처리")
    print("   원인: LLM이 마크다운 코드 펜스나 다른 텍스트를 포함해서 반환")
````
