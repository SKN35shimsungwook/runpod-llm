# MISSION 1. 문자열 메서드 🧹

LLM API 응답에는 불필요한 공백이나 코드펜스가 붙어 오는 경우가 많습니다.  
두 가지 메서드로 깔끔하게 정리할 수 있습니다.

| 메서드 | 역할 | 예시 |
|--------|------|------|
| `.strip()` | 앞뒤 공백·줄바꿈 제거 | `'  텍스트  '.strip()` → `'텍스트'` |
| `.replace(a, b)` | a를 b로 교체 | `'hello'.replace('l', 'r')` → `'herro'` |


````python
# .strip() 테스트
응답1 = '  안녕하세요  '      # 앞뒤 공백
응답2 = '\n답변입니다\n'     # 앞뒤 줄바꿈

print(f'원본: [{응답1}]')
print(f'strip: [{응답1.strip()}]')
print()
print(f'원본: [{응답2}]')
print(f'strip: [{응답2.strip()}]')
````

````python
# .replace() 테스트
응답3 = '```json\n{"score": 4}\n```'   # 코드펜스 포함

print(f'원본: {응답3}')
print()

# 코드펜스 제거 (Day 2~5에서 매일 쓰는 패턴!)
정리됨 = 응답3.replace('```json', '').replace('```', '').strip()
print(f'정리 후: {정리됨}')
````

````python
# 실전 연습 — API 응답 정리하기
지저분한_응답 = '  ```json\n{"result": "성공", "count": 10}\n```  '

# 1단계: 앞뒤 공백 제거
단계1 = 지저분한_응답.strip()
print(f'1단계: {단계1}')

# 2단계: 코드펜스 제거
단계2 = 단계1.replace('```json', '').replace('```', '').strip()
print(f'2단계: {단계2}')
````

# MISSION 2. f-string 기초 📌

f-string = 문자열 앞에 `f`를 붙이고 `{}` 안에 변수를 넣는 방법

```python
이름 = '진환'
f'안녕하세요, {이름}님!'  →  '안녕하세요, 진환님!'
```

**왜 `+` 대신 f-string을 쓰나요?**
```python
# + 연결 (복잡, 실수 많음)
'안녕 ' + 이름 + '님!'

# f-string (읽기 쉬움)
f'안녕 {이름}님!'
```


````python
# f-string 기본
이름 = '노바코프'
점수 = 15
총점 = 20

# 변수 삽입
print(f'회사명: {이름}')

# 계산식도 가능
print(f'점수: {점수}/{총점}')
print(f'비율: {점수/총점*100:.1f}%')  # :.1f = 소수점 1자리
````

````python
# f-string 포맷 실습
비용 = 0.003575
토큰 = 1234567

# 소수점 자리 지정
print(f'비용 (소수점 2자리): ${비용:.2f}')
print(f'비용 (소수점 6자리): ${비용:.6f}')

# 천 단위 쉼표
print(f'토큰 수: {토큰:,}')

# 조건식 (합격/불합격)
print(f'결과: {"합격" if 점수 >= 12 else "불합격"}')
````

### ✅ 결과 확인
- `:.1f` 포맷이 소수점 1자리로 출력됐나요?
- `:,` 가 숫자에 쉼표를 추가했나요?

# MISSION 3. f-string 활용 — 프롬프트 동적 생성 🚀

프롬프트를 매번 손으로 쓰는 대신,  
**변수만 바꾸면** 다른 도메인의 프롬프트가 나옵니다.


````python
# 변수로 프롬프트 만들기
도메인 = '법률'
전문가 = '10년 경력 변호사'
태스크 = '계약서 검토'

프롬프트 = f'당신은 {도메인} 분야의 {전문가}입니다. {태스크}를 정확하게 수행하세요.'
print(프롬프트)
````

````python
# 변수만 바꾸면 완전히 다른 프롬프트!
설정_목록 = [
    ('의료',  '내과 전문의',  '증상 분석'),
    ('세무',  '공인 세무사',  '세금 신고 안내'),
    ('법률',  '노동법 전문가','해고 사건 검토'),
]

for 도메인, 전문가, 태스크 in 설정_목록:
    프롬프트 = f'당신은 {도메인} 분야의 {전문가}입니다. {태스크}를 수행하세요.'
    print(프롬프트)
````

# MISSION 4. 멀티라인 f-string — 긴 프롬프트 작성 📃

실제 프롬프트는 여러 줄입니다.  
삼중 따옴표 `"""` 와 f-string을 함께 씁니다.

```python
프롬프트 = f"""
여러 줄
프롬프트
""".strip()   # ← .strip() 필수! 앞뒤 빈 줄 제거
```


````python
# 멀티라인 f-string
도메인 = '법률'
최대_길이 = '200자'

시스템_프롬프트 = f"""
당신은 {도메인} 분야 전문가 AI입니다.

다음 규칙을 반드시 지키세요:
1. 답변은 {최대_길이} 이내로 작성
2. 불확실한 경우 '전문가 상담 권장' 명시
3. 법조항 번호를 가능한 포함
""".strip()

print(시스템_프롬프트)
print()
print(f'길이: {len(시스템_프롬프트)}자')
````

````python
# .strip() 이 왜 필요한지 비교
without_strip = f"""
첫 번째 줄
두 번째 줄
"""

with_strip = f"""
첫 번째 줄
두 번째 줄
""".strip()

print(f'.strip() 없음: [{without_strip}]')
print(f'.strip() 있음: [{with_strip}]')
````

# MISSION 5. 이중 중괄호 `{{}}` ⚠️

f-string에서 `{변수}` 는 변수 삽입입니다.  
LLM에게 JSON 형식을 알려줄 때는 `{}` 를 **텍스트로** 출력해야 합니다.

```
{}  → 변수 삽입 (에러 or 변수값)
{{}} → 중괄호 문자 그대로 출력
```


````python
# 이중 중괄호 비교
도메인 = '법률'

# ✅ 이중 중괄호 — 텍스트로 출력
프롬프트 = f"""당신은 {도메인} 전문가입니다.

아래 JSON 형식으로만 출력하세요:
{{"score": 1~5, "reason": "근거"}}
"""

print(프롬프트)
````

````python
# 에러 시연 — 단일 중괄호를 쓰면 어떻게 되는가
도메인 = '법률'

try:
    # 이렇게 쓰면 score 변수를 찾으려다 에러!
    잘못된_프롬프트 = f'출력: {"score": 1~5}'
except SyntaxError as e:
    print(f'⚠️ SyntaxError 발생: {e}')
    print('→ JSON 형식에는 반드시 이중 중괄호 {{}} 를 써야 합니다!')
````

# MISSION 6. json.loads() — 문자열 → 딕셔너리 📥

LLM이 JSON 형식으로 응답을 돌려줍니다.  
이 응답은 **문자열**이기 때문에 딕셔너리로 변환해야 값을 꺼낼 수 있습니다.

```python
import json

json.loads('문자열')  →  딕셔너리
```


````python
import json

# LLM이 돌려준 JSON 응답 (문자열)
llm_응답 = '{"score": 4, "reason": "근거가 명확함", "missing": ["판례"]}'

print(f'원본 타입: {type(llm_응답)}')
print(f'원본 값:  {llm_응답}')
````

````python
# json.loads() 로 딕셔너리로 변환
파싱됨 = json.loads(llm_응답)

print(f'변환 후 타입: {type(파싱됨)}')
print()

# 값 꺼내기
print(f'score:   {파싱됨["score"]}')
print(f'reason:  {파싱됨["reason"]}')
print(f'missing: {파싱됨["missing"]}')
print(f'첫 번째: {파싱됨["missing"][0]}')
````

### ✅ 결과 확인
- 원본은 `str` 타입이었다가 `dict` 로 바뀌었나요?
- `파싱됨['score']` 로 값을 꺼낼 수 있나요?

> 💡 `json.loads` 의 `s` = **s**tring (문자열 대상)

# MISSION 7. json.dumps() — 딕셔너리 → 문자열 📤

결과를 파일로 저장하거나 전송할 때는  
딕셔너리를 문자열로 변환합니다.

```python
json.dumps(딕셔너리, ensure_ascii=False, indent=2)
```

> ⚠️ **`ensure_ascii=False` 필수!**  
> 기본값은 True — 한글이 `\uXXXX` 형태로 깨집니다.


````python
# json.dumps() 기본 — 한글 깨짐!
결과 = {'score': 4, 'reason': '근거가 명확함'}

기본 = json.dumps(결과)
print(f'기본값: {기본}')
print('⚠️ 한글이 \\uXXXX 로 깨졌습니다!')
````

````python
# ensure_ascii=False 로 한글 정상 출력
한글정상 = json.dumps(결과, ensure_ascii=False)
print(f'ensure_ascii=False: {한글정상}')
print()

# indent=2 로 보기 좋게
예쁘게 = json.dumps(결과, ensure_ascii=False, indent=2)
print(f'indent=2:')
print(예쁘게)
````

````python
# JSONL 파일 저장 연습
훈련_데이터 = [
    {'instruction': '보증금 반환 절차는?', 'input': '', 'output': '임차권등기명령을 신청하세요.'},
    {'instruction': '해고예고수당이란?',   'input': '', 'output': '30일 전 예고 없이 해고 시 지급합니다.'},
]

# JSONL 형식: 한 줄에 하나의 JSON
print('[JSONL 형식 미리보기]')
for 항목 in 훈련_데이터:
    print(json.dumps(항목, ensure_ascii=False))
````

### ✅ 결과 확인

- `indent=2` 로 보기 좋게 출력됐나요?

> 💡 Day 2~5에서 JSONL 파일을 저장할 때마다  
> `json.dumps(항목, ensure_ascii=False) + '\n'` 을 씁니다!

# MISSION 8. try/except — 에러가 나도 멈추지 않기 🛡️

API 호출은 언제든 실패할 수 있습니다.  
에러가 나도 프로그램이 계속 실행되도록 만드는 것이 `try/except` 입니다.

```python
try:
    # 실패할 수 있는 코드
except Exception as e:
    # 실패했을 때 실행할 코드
```


````python
import json
json.loads('이것은 JSON이 아닙니다')
````

````python
# try/except 없을 때 — 프로그램이 멈춤
print('실행 시작')

try:
    json.loads('이것은 JSON이 아닙니다')  # 에러 발생!
except json.JSONDecodeError as e:
    print(f'⚠️ 파싱 실패: {e}')

print('실행 계속됨!')   # ← try/except 덕분에 여기까지 도달
````

````python
# 안전한 JSON 파싱 함수 (Day 2~5 재사용)
def safe_parse_json(raw):
    '''LLM 응답에서 JSON을 안전하게 추출한다'''
    # 코드펜스 제거
    raw = raw.strip().replace('```json', '').replace('```', '').strip()

    try:
        return json.loads(raw)
    except json.JSONDecodeError:
        print(f'JSON 파싱 실패: {raw[:50]}')
        return None   # None을 반환하면 프로그램은 계속 실행됨


# 테스트
테스트1 = '```json\n{"score": 4, "reason": "좋음"}\n```'  # 코드펜스 특히 Chatgpt나 gemini, claude등에서 많이 보임
테스트2 = '{"score": 5, "reason": "매우 좋음"}'            # 정상
테스트3 = '이것은 JSON이 아닙니다'                          # 실패

print(safe_parse_json(테스트1))
print(safe_parse_json(테스트2))
print(safe_parse_json(테스트3))
````

````python
# 지수 백오프 재시도 함수 (API 실습에서 사용)
import time

def safe_api_call(client, messages, max_retries=3):
    '''API 호출 실패 시 점점 오래 기다린 후 재시도한다'''
    for 시도 in range(max_retries):
        try:
            resp = client.chat.completions.create(
                model='gpt-4o-mini',
                messages=messages
            )
            return resp.choices[0].message.content
        except Exception as e:
            대기 = 2 ** 시도   # 1초 → 2초 → 4초
            print(f'[{시도+1}/{max_retries}] 실패: {e}')
            print(f'  → {대기}초 후 재시도...')
            time.sleep(대기)

    print('❌ 최대 재시도 초과')
    return None


# 재시도 간격 확인
print('[재시도 간격]')
for 시도 in range(3):
    print(f'  {시도+1}번째 실패 후: {2**시도}초 대기')
````

### ✅ 결과 확인
- `safe_parse_json` 이 코드펜스를 제거하고 파싱에 성공했나요?
- 파싱 실패 시 `None` 을 반환하고 프로그램이 계속 실행됐나요?
- 재시도 간격이 1→2→4초로 늘어나나요?

> 💡 **이 두 함수가 Day 2~5의 기반입니다!**
